#Data preparation
All data has to be numerical.

We have to convert the non numerical columns 'Nationality' and 'Go' into numerical values.

Pandas has a map() method that takes a dictionary with information on how to convert the values.

{'UK': 0, 'USA': 1, 'N': 2}

Means convert the values 'UK' to 0, 'USA' to 1, and 'N' to 2.

```python
d = {'UK': 0, 'USA': 1, 'N': 2}
df['Nationality'] = df['Nationality'].map(d)

# What map function does?
# The map() function replaces each value in a column with a corresponding value from a dictionary.
# Here, it converts: 'UK' -> 0, 'USA' -> 1, 'N' -> 2
# This transforms categorical text data into numerical values for machine learning.

# Can you do the same to convert Go to numerical?
go_map = {'YES': 1, 'NO': 0}
df['Go'] = df['Go'].map(go_map)
print("\nAfter mapping Go column:")
print(df)

# Can we use label encoding instead? Try to use it for Go label
from sklearn.preprocessing import LabelEncoder

# Create a copy to demonstrate LabelEncoder
df_copy = pd.read_csv('Lab 3 dataset.csv')
df_copy['Nationality'] = df_copy['Nationality'].map(d)

le = LabelEncoder()
df_copy['Go_encoded'] = le.fit_transform(df_copy['Go'])

print("\nUsing LabelEncoder for Go column:")
print(df_copy)
print("\nLabel mapping:", dict(zip(le.classes_, le.transform(le.classes_))))
print("\nBoth methods work! map() gives you control over the mapping,")
print("while LabelEncoder automatically assigns numbers alphabetically.")
```

#ML Algorithm

#Model training

```python
clf = LogisticRegression()
clf = clf.fit(X.values, y)
```

#Predict Values
We can use the Logistic Regression to predict new values.
*italicised text*
Example: Should I go see a show starring a 40 years old American comedian, with 10 years of experience, and a comedy ranking of 7?

```python
prediction = clf.predict([[40, 10, 7, 1]])
print("Prediction (numerical):", prediction)

# Reflect on the result, what does it mean?
# The result is either 1 or 0:
# - 1 means YES, you should go to the show
# - 0 means NO, you should not go to the show
# The model uses the features (age=40, experience=10, rank=7, nationality=USA)
# to predict whether you would enjoy the show.

# Add if-else to print the result in a text-based format
if prediction[0] == 1:
    print("\nRECOMMENDATION: YES, you should go see this show!")
    print("The model predicts you will enjoy it.")
else:
    print("\nRECOMMENDATION: NO, you should skip this show.")
    print("The model predicts you might not enjoy it.")

# Additional info
print(f"\nPrediction confidence (probability): {clf.predict_proba([[40, 10, 7, 1]])}")
```

```python
# Reflect on the problem:

print("=" * 60)
print("PROBLEM ANALYSIS")
print("=" * 60)

print("\n1. Is it supervised or unsupervised?")
print("   ANSWER: SUPERVISED LEARNING")
print("   Why? Because we have labeled data (the 'Go' column).")
print("   We know the correct answer for each training example.")

print("\n2. Is it classification or regression?")
print("   ANSWER: CLASSIFICATION")
print("   Why? We're predicting discrete categories (YES=1 or NO=0),")
print("   not continuous numerical values.")

print("\n3. Alternative algorithms for this problem:")
print("   From https://scikit-learn.org/stable/supervised_learning.html")

# Try alternative algorithms
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# Using full dataset (X, y) since we have limited data
# Decision Tree
print("\n   a) Decision Tree Classifier:")
dt = DecisionTreeClassifier(random_state=42)
dt.fit(X.values, y)
dt_pred = dt.predict([[40, 10, 7, 1]])
print(f"      Prediction: {dt_pred[0]} ({'Go' if dt_pred[0]==1 else 'No Go'})")

# Random Forest
print("\n   b) Random Forest Classifier:")
rf = RandomForestClassifier(random_state=42)
rf.fit(X.values, y)
rf_pred = rf.predict([[40, 10, 7, 1]])
print(f"      Prediction: {rf_pred[0]} ({'Go' if rf_pred[0]==1 else 'No Go'})")

# Support Vector Machine
print("\n   c) Support Vector Classifier (SVM):")
svm = SVC(random_state=42)
svm.fit(X.values, y)
svm_pred = svm.predict([[40, 10, 7, 1]])
print(f"      Prediction: {svm_pred[0]} ({'Go' if svm_pred[0]==1 else 'No Go'})")

# K-Nearest Neighbors
print("\n   d) K-Nearest Neighbors:")
knn = KNeighborsClassifier()
knn.fit(X.values, y)
knn_pred = knn.predict([[40, 10, 7, 1]])
print(f"      Prediction: {knn_pred[0]} ({'Go' if knn_pred[0]==1 else 'No Go'})")

print("\n" + "=" * 60)
print("All these algorithms are suitable for classification problems!")
print("=" * 60)
```

```python
print(clf.predict_proba([[40, 10, 7, 1]]))

# Reflect on the result and what do these two values represented.
print("\nEXPLANATION OF PREDICT_PROBA OUTPUT:")
print("=" * 50)
print("The output [[0.41174317 0.58825683]] contains TWO probabilities:")
print()
print("  - 0.4117 (41.17%): Probability of class 0 (NO - don't go)")
print("  - 0.5883 (58.83%): Probability of class 1 (YES - go)")
print()
print("The model predicts the class with higher probability.")
print("Since 0.5883 > 0.4117, the prediction is 1 (YES, go to the show).")
print()
print("Note: These probabilities sum to 1.0 (100%)")
print("This is called 'soft prediction' vs 'hard prediction' (just 0 or 1)")
```

```python
# Task 1 Solution: Use Decision Tree and Random Forest algorithms

print("=" * 70)
print("TASK 1: DECISION TREE AND RANDOM FOREST IMPLEMENTATION")
print("=" * 70)

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X.values, y, test_size=0.3, random_state=42)
print(f"\nData split: {len(X_train)} training samples, {len(X_test)} testing samples")

# ===== DECISION TREE ALGORITHM =====
print("\n1. DECISION TREE CLASSIFIER")
print("-" * 70)

# Create and train Decision Tree model
dt_model = DecisionTreeClassifier(
    criterion='gini',  # or 'entropy'
    max_depth=3,       # prevent overfitting on small dataset
    random_state=42
)
dt_model.fit(X_train, y_train)

# Make predictions
dt_predictions = dt_model.predict(X_test)

# Evaluate
print(f"Training Score: {dt_model.score(X_train, y_train):.2f}")
print(f"Testing Score: {dt_model.score(X_test, y_test):.2f}")

# Test prediction
dt_test_pred = dt_model.predict([[40, 10, 7, 1]])
print(f"Prediction for [Age=40, Exp=10, Rank=7, USA]: {'Go' if dt_test_pred[0]==1 else 'No Go'}")

# ===== RANDOM FOREST ALGORITHM =====
print("\n2. RANDOM FOREST CLASSIFIER")
print("-" * 70)

# Create and train Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,  # number of trees
    max_depth=3,
    random_state=42
)
rf_model.fit(X_train, y_train)

# Make predictions
rf_predictions = rf_model.predict(X_test)

# Evaluate
print(f"Training Score: {rf_model.score(X_train, y_train):.2f}")
print(f"Testing Score: {rf_model.score(X_test, y_test):.2f}")

# Test prediction
rf_test_pred = rf_model.predict([[40, 10, 7, 1]])
print(f"Prediction for [Age=40, Exp=10, Rank=7, USA]: {'Go' if rf_test_pred[0]==1 else 'No Go'}")

# Feature importance (Random Forest)
print("\nFeature Importance (Random Forest):")
feature_names = ['Age', 'Experience', 'Rank', 'Nationality']
for name, importance in zip(feature_names, rf_model.feature_importances_):
    print(f"  {name}: {importance:.4f}")

# ===== COMPARISON =====
print("\n3. ALGORITHM COMPARISON")
print("-" * 70)
print("Key Differences:")
print("- Decision Tree: Single tree, easy to interpret, can overfit")
print("- Random Forest: Ensemble of trees, more robust, reduces overfitting")
print("- Logistic Regression: Linear model, outputs probabilities")
print("\n" + "=" * 70)
```
