```python
from sklearn.datasets import load_digits
```

```python
digits = load_digits()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    digits.data, digits.target, random_state=11, test_size=0.20)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
```

```python
estimators = {
        'KNeighborsClassifier': KNeighborsClassifier(),
        'SVC': SVC(gamma='scale'),
        'GaussianNB': GaussianNB()}
```

```python
from sklearn.model_selection import KFold
```

```python
kfold = KFold(n_splits=10, random_state=11, shuffle=True)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
for estimator_name, estimator_object in estimators.items():
      kfold = KFold(n_splits=10, random_state=11, shuffle=True)

      scores = cross_val_score(estimator=estimator_object,X=digits.data, y=digits.target, cv=kfold)

      print(f'{estimator_name:>20}: ' + f'mean accuracy={scores.mean():.2%}; ' + f'standard deviation={scores.std():.2%}')
```

```python
# Try more classifiers and replicate the above steps:
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Add new classifiers to test
additional_estimators = {
    'DecisionTreeClassifier': DecisionTreeClassifier(random_state=11),
    'RandomForestClassifier': RandomForestClassifier(random_state=11)
}

print("Additional Classifiers:")
print("-" * 60)
for estimator_name, estimator_object in additional_estimators.items():
    kfold = KFold(n_splits=10, random_state=11, shuffle=True)
    scores = cross_val_score(estimator=estimator_object, X=digits.data, y=digits.target, cv=kfold)
    print(f'{estimator_name:>25}: ' + f'mean accuracy={scores.mean():.2%}; ' + f'standard deviation={scores.std():.2%}')
```
