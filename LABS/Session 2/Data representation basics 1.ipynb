**What is NumPy?**
NumPy (Numerical Python) is a fundamental Python library for numerical computing. Its primary purpose is to provide support for large, multi-dimensional arrays and matrices, along with a collection of high-level mathematical functions to operate on these arrays efficiently. NumPy is the foundation for most scientific computing in Python.

**What is Pandas?**
Pandas is a powerful Python library for data manipulation and analysis. Its primary purpose is to provide easy-to-use data structures (Series and DataFrame) and data analysis tools for working with structured data. Pandas is built on top of NumPy and is essential for data cleaning, transformation, and analysis tasks.

```python
# Import important libraray
import numpy as np
import pandas as pd

# Two data types: Series and DataFrame
# Series can only contain single list with index
# Dataframe is a collection of series that can be used to analyse the data (a table with rows and columns)
from pandas import Series, DataFrame
```

```python
# Without an index
#serios_obj = Series(np.arange(10))
#serios_obj
# With an index
# Practise both
series_obj = Series(np.arange(10), index=['row 1', 'row 2', 'row 3', 'row 4', 'row 5', 'row 6','row 7', 'row 8', 'row 9', 'row 10'])
series_obj
```

```python
# Print the index of row 7
series_obj['row 7']
```

```python
# Print both indexes at row 1 & 8
series_obj[[0, 7]]
```

```python
# Populate the list with random values
np.random.seed(25)

# Search random.seed in Python and provide an explaniation here:
# The np.random.seed() function sets the random number generator to a specific starting point.
# This ensures reproducibility - running the code multiple times will generate the same 'random' numbers.
# This is useful for debugging and ensuring consistent results in experiments.

# What does seed function do?
# It initializes the random number generator with a specific value (25 in this case).
# Without a seed, you get different random numbers each time you run the code.
# With a seed, the sequence of 'random' numbers is predictable and repeatable.

DF_obj = DataFrame(np.random.rand(36).reshape((6,6)),
                   index=['row 1', 'row 2', 'row 3', 'row 4','row 5','row 6'],
                   columns=['column 1','column 2','column 3','column 4','column 5','column 6'])
DF_obj

# Reflection on difference with and without np.random.seed:
# WITH seed: The DataFrame values are identical every time you run the code.
# WITHOUT seed: The DataFrame contains different random values each time you run it.
# Using a seed is important for reproducible research and consistent testing.
```

```python
#Find data at different location
DF_obj.loc[['row 2', 'row 5'], ['column 5', 'column 2']]

# What does loc function do?
# The loc function is used for label-based indexing in Pandas.
# It selects data by row and column LABELS (names).
# In this example: loc[row_labels, column_labels]
# - Rows: 'row 2' and 'row 5'
# - Columns: 'column 5' and 'column 2'
# It returns the intersection of these rows and columns as a DataFrame.
```

### Data slicing
You can use slicing to select and return a slice of several values from a data set. Slicing uses index values so you can use the same square brackets when doing data slicing.

How slicing differs, however, is that with slicing you pass in two index values that are separated by a colon. The index value on the left side of the colon should be the first value you want to select. On the right side of the colon, you write the index value for the last value you want to retrieve. When you execute the code, the indexer then simply finds the first record and the last record and returns every record in between them.

```python
series_obj['row 3': 'row 7']
```

### Comparing with scalars
Now we're going to talk about comparison operators and scalar values. Just in case you don't know that a scalar value is, it's basically just a single numerical value. You can use comparison operators like greater than or less than to return true/false values for all records to indicate how each element compares to a scalar value.

```python
DF_obj < .2

# Reflection on using greater than (>) instead:
# With '<': Returns True for all DataFrame elements less than 0.2, False otherwise.
# With '>': Returns True for all DataFrame elements greater than 0.2, False otherwise.
# This creates a Boolean DataFrame that can be used for filtering data.
# The result shows the logical structure of your data relative to the threshold.
```

### Filtering with scalars

```python
# Retrieves all elements in series_obj that are greater than 6.
series_obj[series_obj > 6]

# Retrive all elements that are smaller than 5
series_obj[series_obj < 5]
```

### Setting values with scalars

```python
# Assign the value 8 to specific rows using .loc
series_obj.loc[['row 1', 'row 5', 'row 8']] = 8
series_obj
```
